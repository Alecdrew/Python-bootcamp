```python
import matplotlib.pyplot as plt

# Extract the columns we want to plot from the dataframe
# Doing this first makes the plotting code easier to read
x = df['W']
y = df['Y']

# Create a Figure and an Axes object
# Figure = the entire canvas
# Axes = the individual plotting area inside the figure
fig, ax = plt.subplots()

# Plot the data on the axes
# We call methods on the ax object instead of using plt.scatter()
# This is the recommended "object-oriented" matplotlib style
ax.scatter(x, y)

# Set axis labels and title using ax methods
ax.set_xlabel('W')
ax.set_ylabel('Y')
ax.set_title('W vs Y')

# Always show the plot explicitly when writing scripts
plt.show()
```

```python
#now we see basic info we can start to enrich the plot with more features such as color, gridlines, size, trendline, data labels, legend etc 
#you can even use two rows of data on the same axis to compare them by using different colors and markers
#we can do all of this in one line of code or we can do it step by step as we did above

x = df['W']
y1 = df['Y']
y2 = df['X']

#create a new figure and axes for the new chart
fig, ax = plt.subplots()

#plot both datasets on the same axes so we can compare them
ax.scatter(x,y1, color='blue', marker='o', label='Y')
ax.scatter(x,y2, color='red', marker='.', label='X')

#add labels and title
ax.set_xlabel('W')
ax.set_ylabel('Values')
ax.set_title('W vs Y and X')

#show legend based on the labels defined above
ax.legend(loc='upper left')

#gridlines make it easier to visually compare values
ax.grid(True)

plt.show()


#the full types of markers can be found in the matplotlib documentation: https://matplotlib.org/stable/api/markers_api.html#module-matplotlib.markers
#the full list of colors can be found in the matplotlib documentation: https://matplotlib.org/stable/gallery/color/named_colors.html
#the full list of plot types can be found in the matplotlib documentation: https://matplotlib.org/stable/plot_types/index.html
#the full list of plot features can be found in the matplotlib documentation: https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.scatter.html#matplotlib.pyplot.scatter

```

```python
#now we can be smart to explore the relationship between the data sets by calculating then plotting the distance between the points
#we can create a for loop to calculate the distance between the points and then plot the distance as a new scatter plot with text to show the distance values

fig, ax = plt.subplots()

#plot the two datasets
ax.scatter(df['W'], df['Y'], color='blue', marker='o', label='Y')
ax.scatter(df['W'], df['X'], color='red', marker='.', label='X')

#zip allows us to loop through multiple columns at the same time
for x, y1, y2 in zip(df['W'], df['Y'], df['X']):
    
    #calculate the vertical distance between the two points
    d = abs(y1 - y2)

    #draw a vertical dashed line between the two values
    ax.plot([x, x], [y1, y2], color='gray', linestyle='--')

    #add text showing the distance value
    #the label is placed halfway between the two points
    ax.text(x, (y1 + y2) / 2, f'{d:.2f}', color='black', fontsize=8, ha='left') 


ax.set_xlabel('W')
ax.set_ylabel('Values')
ax.set_title('W vs Y and X with Distance')
ax.legend(loc='upper left')
ax.grid(True)

plt.show()


#we can also use the distance values to create a new column in the dataframe
#this uses pandas vectorized operations instead of loops which is faster and cleaner

df['Distance_Y_X'] = abs(df['Y'] - df['X'])
df

```

```python
#distribution plots show the distribution of a single variable or the relationship between two variables
import matplotlib.pyplot as plt
import seaborn as sns
tips_df = sns.load_dataset('tips')
sns.set_style('darkgrid') #set the style of the plot to have a dark background with gridlines

#kind could also be 'scatter', 'reg', 'resid', 'kde', 'hex' etc depending on the type of plot you want to create
sns.jointplot(x='total_bill', y='tip', data=tips_df, kind='hex')
```

```python
# pairplot creates a grid of scatterplots for each pair of variables in the dataframe, 
# colored by the 'sex' column. you can add trendlines to the pairplot by setting kind='reg' which will add a regression line to each scatterplot in the grid. 
# you can also specify different types of plots for the diagonal and off-diagonal plots using the diag_kind and kind parameters respectively. for example, 
# diag_kind='kde' will show kernel density estimates on the diagonal instead of histograms, and kind='scatter' will show scatterplots on the off-diagonal instead of regression plots.

sns.pairplot(tips_df, hue='sex', diag_kind='kde', kind='reg') 
```

```python
#catergorical plots show the relationship between a categorical variable and a continuous variable, or between two categorical variables.
#box plots show the distribution of a continuous variable across different categories, and can be used to compare the central tendency and variability of the data across groups.
#violin plots are similar to box plots but also show the kernel density estimate of the distribution, which can provide more information about the shape of the data distribution across categories.

#ignore this line for now, we will come back to it after we learn about box plots and violin plots
fig, ax = plt.subplots(1, 2, figsize=(12,6))


sns.violinplot(x='day', 
               y='tip', 
               data=tips_df, 
               hue='smoker', 
               split=True,
               inner='quartiles',
               palette='Set3',
               ax=ax[0])

# were split=True will split the violin plot into two halves based on the hue variable (in this case, smoker vs non-smoker), 
# as a result its best to update the inner parameter to show the quartiles instead of the full distribution to make it easier to compare the two halves of the plot.
# you could also create a seperate box plot if the lines are too close togerther to compare the two halves of the violin plot
# we can combine with plt to create two subplots to compare the violen with quartiles vs the box plot with the full distribution


sns.violinplot(x='day', 
               y='tip', 
               data=tips_df, 
               hue='smoker', 
               split=True,
               inner=None,
               palette='Set2',
               ax=ax[1])

sns.boxplot(x='day',
            y='tip',
            data=tips_df,
            hue='smoker',
            width=0.2,
            palette='Set2',
            legend=False,
            ax=ax[1])
```

```python
# matrix plots are slightly different instead of having a category and some numerical vaules we have two categorical variables and we want to see the relationship between them.
# we also need to make sure that the data is in the correct format for a matrix plot which typically means we need to pivot the data to create a matrix of values that can be plotted as a heatmap or clustermap.

df_tips = sns.load_dataset('tips')
df_flights = sns.load_dataset('flights')

#dont forget numeric_only=True when calculating correlations for a heatmap to avoid errors with non-numeric columns
tc = df_tips.corr(numeric_only=True)

sns.heatmap(tc, annot=True, cmap='coolwarm')
```

```python
#for flight data we need to pivot
flights_pivot = df_flights.pivot(index='month', columns='year', values='passengers')

# we can use linecolor and width to make the gridlines more visible and we can use cmap to change the color scheme of the heatmap
# we can also use the clusermap function to perform hierarchical clustering on the data and show the clusters in the heatmap, 
# which can help to identify patterns and relationships in the data that may not be immediately apparent from the raw values alone.

plt.figure(figsize=(12,6))
sns.heatmap(
    flights_pivot,
    annot=False,
    cmap='magma',
    linecolor='white',
    linewidths=0.5
)
plt.show()

sns.clustermap(
    flights_pivot,
    annot=False,
    cmap='coolwarm',
    standard_scale=1
)
```

```python
# grids are another way to look at lots of data like subplots but they are more flexible and powerful because they allow you to define the type of plot for each cell in the grid 
# they also allow you to share axes across multiple plots which can make it easier to compare data across different plots.

#you can use it similar to pairplot 

df_iris = sns.load_dataset('iris')

g = sns.PairGrid(df_iris, hue= 'species')
g.map_diag(sns.histplot)
g.map_lower(sns.kdeplot)
g.map_upper(sns.regplot, scatter_kws={'s': 20}, line_kws={'color': 'red'})
```

```python
#unstack is great for creating pivot tables from groupby objects which can then be plotted as heatmaps or clustermaps
# to visualize the relationships between the groups and the values in the dataframe.
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# ── Sample Data ───────────────────────────────────────────────
np.random.seed(42)
df = pd.DataFrame({
    'Date': pd.date_range('2024-01-01', periods=200, freq='D').repeat(3),
    'Reason': np.tile(['EMS', 'Traffic', 'Fire'], 200),
    'Count': np.random.randint(10, 100, 600)
})
df['Day of Week'] = df['Date'].dt.day_name().str[:3]

# ── Step 1: groupby creates a MultiIndex Series (stacked) ─────
grouped = df.groupby(['Day of Week', 'Reason'])['Count'].sum()
print("GROUPED (stacked MultiIndex Series):")
print(grouped.head(5))

# ── Step 2: unstack pivots inner index into columns ───────────
pivot = grouped.unstack()
print("\nUNSTACKED (pivot table — Reason becomes columns):")
print(pivot.head(5))

# ── Step 3: stack() reverses unstack — back to MultiIndex ─────
restacked = pivot.stack()
print("\nRESTACKED (back to MultiIndex Series):")
print(restacked.head(5))

day_order = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
pivot = pivot.reindex(day_order)

# ── Heatmap ───────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(7, 5))
sns.heatmap(pivot, annot=True, fmt='g', cmap='YlOrRd', ax=ax)
ax.set_title('Heatmap — Calls by Day & Reason')
plt.tight_layout()
plt.show()

# ── Clustermap ────────────────────────────────────────────────
sns.clustermap(pivot, annot=True, fmt='g', cmap='YlOrRd', figsize=(7,7))
plt.show()
```

```python
# ============================================================
# MULTI-LEVEL INDEXING (MultiIndex) IN PANDAS
# ============================================================
import pandas as pd
import numpy as np

# ── What is a MultiIndex? ────────────────────────────────────
# A MultiIndex allows multiple levels of indexing on rows OR columns.
# Think of it as a hierarchy of labels — like a folder structure.
# Level 0 = outer (e.g. Category), Level 1 = inner (e.g. Metric)

arrays = [['A', 'A', 'B', 'B'], ['x', 'y', 'x', 'y']]
mi = pd.MultiIndex.from_arrays(arrays, names=['Category', 'Metric'])

df = pd.DataFrame(np.random.randint(1, 100, (4, 4)), columns=mi)
print("MultiIndex DataFrame:")
print(df)

# ── Accessing data ────────────────────────────────────────────
print("\nOuter level slice df['A']:")
print(df['A'])

print("\nInner level slice df['A']['x']:")
print(df['A']['x'])

# ── .xs() — cross section (slice any level) ───────────────────
# Useful when you want to slice on an inner level without the outer
print("\n.xs() — all 'x' columns across categories:")
print(df.xs('x', axis=1, level='Metric'))

# ── swaplevel() — swap level order ───────────────────────────
print("\nswaplevel() — swap Category and Metric:")
print(df.swaplevel(axis=1).sort_index(axis=1))

# ── droplevel() — remove a level ─────────────────────────────
print("\ndroplevel() — remove outer level:")
print(df.droplevel(0, axis=1))

# ── stack() and unstack() ─────────────────────────────────────
# unstack() pivots an inner row index level out into columns
# stack()   is the inverse — collapses columns back into rows

simple = pd.DataFrame({
    'Category': ['A', 'A', 'B', 'B'],
    'Metric':   ['x', 'y', 'x', 'y'],
    'Value':    [10, 20, 30, 40]
})
pivot = simple.groupby(['Category', 'Metric'])['Value'].sum().unstack()
print("\nGroupby + unstack() -> pivot table:")
print(pivot)

print("\nstack() -> back to MultiIndex Series:")
print(pivot.stack())

# ── Renaming levels ───────────────────────────────────────────
df.columns.names = ['Category', 'Metric']
print("\nColumn level names:", df.columns.names)

# ── Summary ───────────────────────────────────────────────────
# .xs()         — cross section slice on any level
# swaplevel()   — swap the order of levels
# droplevel()   — remove an unwanted level
# stack()       — columns -> rows
# unstack()     — rows -> columns (great for pivot tables + heatmaps)
# sort_index()  — always sort after swapping/reindexing
```
